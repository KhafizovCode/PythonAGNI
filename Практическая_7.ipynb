Практическая номер 7

```python
import math
from abc import ABC, abstractmethod
class Shape(ABC):
    name = "Фигура"
    @abstractmethod
    def area(self) -> float:
        pass
    @abstractmethod
    def params(self) -> str:
        pass
    def __str__(self) -> str:
        return self.name
    def __eq__(self, other) -> bool:
        if not isinstance(other, Shape):
            return NotImplemented
        return math.isclose(self.area(), other.area())
    def __hash__(self) -> int:
        return hash(round(self.area(), 9))
    def compare(self, other: "Shape") -> str:
        if self == other:
            return f"Площадь {self} равна площади {other}"
        if self.area() > other.area():
            return f"Площадь {self} больше площади {other}"
        return f"Площадь {self} меньше площади {other}"
class Circle(Shape):
    name = "Круг"
    def __init__(self, radius: float):
        if radius <= 0:
            raise ValueError("Радиус должен быть положительным")
        self.radius = radius
    def area(self) -> float:
        return math.pi * self.radius ** 2
    def params(self) -> str:
        return f"радиус = {self.radius}"
class Rectangle(Shape):
    name = "Прямоугольник"
    def __init__(self, width: float, height: float):
        if width <= 0 or height <= 0:
            raise ValueError("Стороны должны быть положительными")
        self.width = width
        self.height = height
    def area(self) -> float:
        return self.width * self.height
    def params(self) -> str:
        return f"ширина = {self.width}, высота = {self.height}"
if __name__ == "__main__":
    circle = Circle(3)
    rect = Rectangle(4, 7)
    square = Rectangle(5, 5)
    rect2 = Rectangle(5, 5)
    for shape in (circle, rect, square):
        print(f"{shape}: {shape.params()}, площадь = {shape.area():.2f}")
    print(circle.compare(rect))
    print(rect.compare(circle))
    print(square.compare(rect2))
    print("square == rect2:", square == rect2)
    print("circle == rect:", circle == rect)
```
